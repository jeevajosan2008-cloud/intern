# Task 2: Data Visualization with Matplotlib & Seaborn
**Dataset:** Titanic  
**Goal:** Create informative charts: bar charts, scatter plots, histograms, box plots, heatmaps and correlation matrices, with custom palettes and formatting.

## 1. Setup and load data

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs('images', exist_ok=True)

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Basic cleaning (same as Task 1)
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop(columns=['Cabin'])
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['Sex'] = df['Sex'].str.capitalize()
df['Survived_Label'] = df['Survived'].map({0: 'Died', 1: 'Survived'})
df.head()
```

## 2. Custom styling and color palettes

```python
sns.set_theme(style='whitegrid', context='notebook', font_scale=1.1)

palette_surv = {'Died': '#E4572E', 'Survived': '#2E86AB'}
palette_sex = {'Male': '#4C78A8', 'Female': '#F58518'}
palette_class = sns.color_palette('viridis', 3)

plt.rcParams['figure.figsize'] = (8, 5)
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['axes.titlesize'] = 14
```

## 3. Bar charts

```python
fig, ax = plt.subplots()
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', palette=palette_sex, errorbar=None, ax=ax)
ax.set_title('Survival Rate by Passenger Class and Sex')
ax.set_xlabel('Passenger Class')
ax.set_ylabel('Survival Rate')
ax.set_ylim(0, 1)
for c in ax.containers:
    ax.bar_label(c, fmt='%.2f', padding=2)
plt.tight_layout()
plt.savefig('images/bar_survival_class_sex.png', dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots()
sns.countplot(data=df, x='Embarked', hue='Survived_Label', palette=palette_surv, ax=ax)
ax.set_title('Passenger Count by Port of Embarkation')
ax.set_xlabel('Port (C = Cherbourg, Q = Queenstown, S = Southampton)')
ax.set_ylabel('Count')
plt.tight_layout()
plt.savefig('images/bar_embarked_count.png', dpi=150)
plt.show()
```

## 4. Histograms

```python
fig, ax = plt.subplots()
sns.histplot(data=df, x='Age', hue='Survived_Label', bins=30, kde=True,
             palette=palette_surv, alpha=0.6, ax=ax)
ax.set_title('Age Distribution by Survival')
ax.set_xlabel('Age (years)')
plt.tight_layout()
plt.savefig('images/hist_age.png', dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots()
sns.histplot(data=df, x='Fare', bins=40, color='#6A4C93', ax=ax)
ax.set_title('Fare Distribution')
ax.set_xlabel('Fare')
plt.tight_layout()
plt.savefig('images/hist_fare.png', dpi=150)
plt.show()
```

## 5. Scatter plots

```python
fig, ax = plt.subplots(figsize=(9, 6))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived_Label', style='Pclass',
                palette=palette_surv, alpha=0.7, s=70, ax=ax)
ax.set_title('Age vs Fare (colored by survival)')
ax.set_xlabel('Age (years)')
ax.set_ylabel('Fare')
ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.savefig('images/scatter_age_fare.png', dpi=150)
plt.show()
```

## 6. Box plots

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.boxplot(data=df, x='Pclass', y='Age', hue='Pclass', palette=palette_class, legend=False, ax=axes[0])
axes[0].set_title('Age by Passenger Class')
axes[0].set_xlabel('Passenger Class')

sns.boxplot(data=df, x='Pclass', y='Fare', hue='Pclass', palette=palette_class, legend=False, ax=axes[1])
axes[1].set_title('Fare by Passenger Class')
axes[1].set_xlabel('Passenger Class')

plt.tight_layout()
plt.savefig('images/boxplots.png', dpi=150)
plt.show()
```

## 7. Correlation matrix and heatmap

```python
num_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'FamilySize']
corr = df[num_cols].corr()

fig, ax = plt.subplots(figsize=(9, 7))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, linewidths=0.5, cbar_kws={'label': 'Correlation'}, ax=ax)
ax.set_title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('images/correlation_heatmap.png', dpi=150)
plt.show()
```

```python
pivot = df.pivot_table(values='Survived', index='Pclass', columns='Sex', aggfunc='mean')

fig, ax = plt.subplots(figsize=(6, 4))
sns.heatmap(pivot, annot=True, fmt='.2f', cmap='YlGnBu', linewidths=0.5, ax=ax)
ax.set_title('Survival Rate: Class vs Sex')
plt.tight_layout()
plt.savefig('images/heatmap_class_sex.png', dpi=150)
plt.show()
```

## 8. Pair plot

```python
g = sns.pairplot(df[['Survived_Label', 'Age', 'Fare', 'FamilySize']], hue='Survived_Label',
                 palette=palette_surv, corner=True)
g.fig.suptitle('Pair Plot of Key Features', y=1.02)
g.savefig('images/pairplot.png', dpi=120)
plt.show()
```

## 9. Key Takeaways
- Women and 1st-class passengers had much higher survival rates (bar chart and heatmap).
- Younger passengers, especially children, were more likely to survive (age histogram).
- Fare is heavily right-skewed; 1st class pays far more (histogram and box plot).
- Pclass and Fare show the strongest correlation with Survived, in opposite directions (heatmap).
